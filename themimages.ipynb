```python
import os

def count_files_in_directory(root_dir):
    file_count = 0
    for root, dirs, files in os.walk(root_dir):
        file_count += len(files)
    
    return file_count

folder_path = 'train/train'

if os.path.exists(folder_path):
    total = count_files_in_directory(folder_path)
    print(f"Total files found: {total}")
else:
    print("The specified directory does not exist.")
```

```python
import shutil
import random
from pathlib import Path

def create_validation_set(train_dir, val_dir, split_ratio=0.1):
    train_path = Path(train_dir)
    val_path = Path(val_dir)

    categories = [d for d in train_path.iterdir() if d.is_dir()]

    for category in categories:
        category_val_path = val_path / category.name
        category_val_path.mkdir(parents=True, exist_ok=True)

        files = [f for f in category.iterdir() if f.is_file()]
        
        num_to_move = int(len(files) * split_ratio)
        files_to_move = random.sample(files, num_to_move)

        for file in files_to_move:
            shutil.move(str(file), str(category_val_path / file.name))
        
        print(f"Moved {num_to_move} files from {category.name} to validation.")

# Usage
create_validation_set('train/train', 'val/val')
```
